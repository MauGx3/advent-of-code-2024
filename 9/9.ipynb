Let's break this down into a clear approach:

1. First, we need to find all the positions of dots in our string since these are the positions where we'll be moving digits to.
2. For each dot position, we need to:
   - Find the rightmost digit in the string
   - Move that digit to the dot position
   - Replace the original digit position with a dot

Let's go through how this works:

1. We convert the string to a list because strings in Python are immutable, and we need to make changes to individual positions.

2. The main while loop continues as long as there exists a dot that has at least one digit to its right. This condition is checked using a nested comprehension that looks for:
   - Any position containing a dot
   - That has at least one digit in a later position

3. For each iteration:
   - We find the leftmost dot that still has digits to its right
   - We find the rightmost digit in the string
   - We swap these positions, effectively moving the digit left and the dot right

4. The process repeats until all digits are on the left and all dots are on the right

````python
def move_digits_left(s: str) -> str:
    # Convert to list for easier manipulation since strings are immutable
    chars = list(s)

    # Keep going until we've processed all dots
    # We can identify this by checking if there are any dots before digits
    while any(
        c == "." and any(d.isdigit() and i < j for j, d in enumerate(chars))
        for i, c in enumerate(chars)
    ):
        # Find the first dot that has digits to its right
        dot_pos = next(
            i
            for i, c in enumerate(chars)
            if c == "." and any(d.isdigit() for d in chars[i:])
        )

        # Find the rightmost digit
        digit_pos = len(chars) - 1
        while digit_pos >= 0 and not chars[digit_pos].isdigit():
            digit_pos -= 1

        # Perform the swap
        chars[dot_pos] = chars[digit_pos]
        chars[digit_pos] = "."

    return "".join(chars)


# Test the function
test_str = "00...111...2...333.44.5555.6666.777.888899"
result = move_digits_left(test_str)
print(result)
print(example_transformation[-1])
````

Let's analyze what makes this algorithm work:

1. **Greedy Approach**: The algorithm uses a greedy strategy, meaning it makes the locally optimal choice at each step. In this case:
   - It always looks for the leftmost dot that still has digits to its right
   - It always moves the rightmost available digit to fill that dot

2. **Invariants**: The algorithm maintains several important properties throughout its execution:
   - All digits to the left of the current dot position remain in their original relative order
   - The total number of dots and digits stays constant
   - After each step, the rightmost portion of the string has more dots and fewer digits

3. **Termination Condition**: The algorithm stops when there are no more dots that have digits to their right. This is guaranteed to happen because:
   - Each step moves one digit left
   - Each step moves one dot right
   - The number of possible moves is finite

Let's see it in action with the example "12..3.4":

Key characteristics of the algorithm:

1. **Time Complexity**: 
   - Worst case: O(n²) where n is the length of the string
   - This is because for each dot (potentially O(n) dots), we might need to scan the entire string to find digits

2. **Space Complexity**:
   - O(n) to store the character array
   - The space is modified in-place after the initial conversion

3. **Stability**:
   - The algorithm is stable with respect to the relative order of digits
   - Digits that start on the left stay on the left
   - The relative order of digits is preserved throughout the transformation

4. **Correctness Properties**:
   - It always terminates (the number of possible moves is finite)
   - It preserves the total count of digits and dots
   - It maintains the relative ordering of digits

Understanding these properties helps us know when and how to use this algorithm:
- It's good for smaller strings where O(n²) performance is acceptable
- It's particularly useful when maintaining digit order is important
- It's easy to visualize and debug due to its step-by-step nature

Let me help you understand how we can optimize this algorithm for large strings. The key insight is that our current approach scans the string multiple times, which becomes inefficient as the string grows. Let's explore a more efficient solution.

Here's an optimized approach that only needs to scan the string once:

```python
def optimize_digit_movement(s: str, verbose=True) -> str:
    # Convert string to list for manipulation
    chars = list(s)
    n = len(chars)
    
    # First pass: collect positions of dots and digits
    dot_positions = []
    digit_positions = []
    
    for i in range(n):
        if chars[i] == '.':
            dot_positions.append(i)
        elif chars[i].isdigit():
            digit_positions.append(i)
            
    # Sort our positions to ensure we process from left to right for dots
    # and right to left for digits
    dot_positions.sort()  # dots from left to right
    digit_positions.sort(reverse=True)  # digits from right to left
    
    # Create result string in one go
    result = [''] * n
    
    # Fill in all non-dot and non-digit characters first
    for i in range(n):
        if chars[i] != '.' and not chars[i].isdigit():
            result[i] = chars[i]
    
    # Now handle the movement of digits to dot positions
    dot_idx = 0  # Index in dot_positions
    digit_idx = 0  # Index in digit_positions
    
    # Move digits to earliest available dots
    while dot_idx < len(dot_positions) and digit_idx < len(digit_positions):
        dot_pos = dot_positions[dot_idx]
        digit_pos = digit_positions[digit_idx]
        
        # Only move digit if it's to the right of the dot
        if digit_pos > dot_pos:
            if verbose:
                print(f"Moving digit '{chars[digit_pos]}' from position {digit_pos} to {dot_pos}")
            result[dot_pos] = chars[digit_pos]
            result[digit_pos] = '.'
        else:
            # If digit is already to the left of dot, keep it in place
            result[digit_pos] = chars[digit_pos]
            result[dot_pos] = '.'
            
        dot_idx += 1
        digit_idx += 1
    
    # Fill in any remaining positions
    while digit_idx < len(digit_positions):
        pos = digit_positions[digit_idx]
        result[pos] = chars[pos]
        digit_idx += 1
        
    while dot_idx < len(dot_positions):
        pos = dot_positions[dot_idx]
        result[pos] = '.'
        dot_idx += 1
    
    return ''.join(result)

# Test the optimized function with a larger string
test_str = '00...111...2...333.44.5555.6666.777.888899'
print("Initial string:", test_str)
result = optimize_digit_movement(test_str)
print("Final result:", result)
```

Let's understand why this approach is more efficient:

1. **Time Complexity Improvement**:
   - The original algorithm: O(n²)
   - This optimized version: O(n)
   - We only need one initial pass through the string to collect positions
   - After that, we work with sorted lists of positions

2. **Memory Trade-off**:
   - We use a bit more memory to store the positions of dots and digits
   - This extra space is O(n) in the worst case
   - The trade-off is worth it for the significant speed improvement

3. **Key Optimizations**:
   - No repeated scanning of the string
   - Direct mapping of positions
   - Single-pass construction of the result

````python
def calculate_checksum(compacted_file: str) -> int:
    """Calculates a checksum for a compacted file by summing the product of each digit and its position.

    Args:
        compacted_file (str): A string representing the compacted file content.

    Returns:
        int: The calculated checksum value.

    Example:
        >>> calculate_checksum("a1b2c3")
        8  # 1*1 + 2*3 + 3*5 = 8
    """
    # Initialize the checksum to 0
    checksum = 0

    # Iterate over the characters in the compacted file
    for i, c in enumerate(compacted_file):
        # If the character is a digit, add its value to the checksum
        if c.isdigit():
            checksum += int(c) * i

    return checksum
````

````python
def sort_blocks(s: list):
    # Get positions of dots/free space
    free_space_indexes = [i for i, x in enumerate(s) if x == "."]

    # Get file blocks (integers)
    file_blocks = [x for x in s if isinstance(x, int)]
    file_blocks_copy = file_blocks.copy()

    # Working copy of input list
    sorted_blocks = s.copy()

    # Process each move
    for idx, file_block in zip(free_space_indexes, file_blocks):
        # Move block to free space
        sorted_blocks[idx] = file_blocks_copy[-1]
        # Remove moved block
        file_blocks_copy = file_blocks_copy[:-1]
        # Add dot at end
        sorted_blocks = sorted_blocks[:-1] + ["."]

        print(f"Current state: {sorted_blocks}")

    # Calculate checksum
    checksum = 0
    for pos, val in enumerate(sorted_blocks):
        if isinstance(val, int):
            checksum += pos * val

    return checksum
````

````python
def optimize_digit_movement(s: str, verbose=True) -> str:
    """Optimizes the movement of digits in a string by moving them to the leftmost available dot positions.

    This function takes a string containing digits, dots, and other characters, and moves digits
    to dot positions that occur earlier in the string (to the left). The movement preserves the
    relative order of digits and only moves digits to the left if there's an available dot position.

    Args:
        s (str): The input string containing digits (0-9), dots (.), and other characters.
        verbose (bool, optional): If True, prints details about digit movements. Defaults to True.

    Returns:
        str: A new string with digits moved to the leftmost possible dot positions.

    Example:
        >>> optimize_digit_movement("1.2.3", verbose=False)
        "123.."
        >>> optimize_digit_movement(".1.2", verbose=False)
        "12.."
        >>> optimize_digit_movement("..12", verbose=False)
        "12.."
    """
    # Convert string to list for manipulation
    chars = list(s)
    n = len(chars)

    # First pass: collect positions of dots and digits
    dot_positions = []
    digit_positions = []

    for i in range(n):
        if chars[i] == ".":
            dot_positions.append(i)
        elif chars[i].isdigit():
            digit_positions.append(i)

    # Sort our positions to ensure we process from left to right for dots
    # and right to left for digits
    dot_positions.sort()  # dots from left to right
    digit_positions.sort(reverse=True)  # digits from right to left

    # Create result string in one go
    result = [""] * n

    # Fill in all non-dot and non-digit characters first
    for i in range(n):
        if chars[i] != "." and not chars[i].isdigit():
            result[i] = chars[i]

    # Now handle the movement of digits to dot positions
    dot_idx = 0  # Index in dot_positions
    digit_idx = 0  # Index in digit_positions

    # Move digits to earliest available dots
    while dot_idx < len(dot_positions) and digit_idx < len(digit_positions):
        dot_pos = dot_positions[dot_idx]
        digit_pos = digit_positions[digit_idx]

        # Only move digit if it's to the right of the dot
        if digit_pos > dot_pos:
            if verbose:
                print(
                    f"Moving digit '{chars[digit_pos]}' from position {digit_pos} to {dot_pos}"
                )
            result[dot_pos] = chars[digit_pos]
            result[digit_pos] = "."
        else:
            # If digit is already to the left of dot, keep it in place
            result[digit_pos] = chars[digit_pos]
            result[dot_pos] = "."

        dot_idx += 1
        digit_idx += 1

    # Fill in any remaining positions
    while digit_idx < len(digit_positions):
        pos = digit_positions[digit_idx]
        result[pos] = chars[pos]
        digit_idx += 1

    while dot_idx < len(dot_positions):
        pos = dot_positions[dot_idx]
        result[pos] = "."
        dot_idx += 1

    return "".join(result)
````

````python
from dataclasses import dataclass
from datetime import datetime
from typing import Optional


@dataclass
class BlockGroup:
    file_id: Optional[int] = None
    size: int = 0
    is_free_space: bool = False


@dataclass
class Block:
    file_id: Optional[int] = None
    is_free_space: bool = False

    def __repr__(self):
        if self.is_free_space:
            return "."
        return str(self.file_id)


def get_blocks(disk_map: [BlockGroup]):
    blocks = []
    for block_group in disk_map:
        for i in range(block_group.size):
            blocks.append(Block(block_group.file_id, block_group.is_free_space))
    return blocks


def get_checksum(blocks: [Block]):
    return sum(
        [
            block.file_id * i
            for i, block in enumerate(blocks)
            if block.file_id is not None
        ]
    )


def get_disk_map():
    disk_map: [BlockGroup] = []
    with open("input.txt") as file:
        line = file.readline().strip()
        for i in range(0, len(line), 2):
            disk_map.append(BlockGroup(int(i / 2), int(line[i]), False))
            if i < len(line) - 1:
                disk_map.append(BlockGroup(None, int(line[i + 1]), True))
    return disk_map


# Part 1

blocks = get_blocks(get_disk_map())
print(blocks)

previous_free_block_id = 0
for i in range(len(blocks) - 1, -1, -1):
    if not blocks[i].is_free_space:
        for j in range(previous_free_block_id, i):
            if blocks[j].is_free_space:
                blocks[i], blocks[j] = blocks[j], blocks[i]
                previous_free_block_id = j
                break

checksum = get_checksum(blocks)

print("Part 1 Checksum:", checksum)

# Part 2

start_time = datetime.now()

block_groups = get_disk_map()

i = len(block_groups) - 1

while i >= 0:
    if block_groups[i].file_id is not None:
        for j in range(i):
            if (
                block_groups[j].is_free_space
                and block_groups[j].size >= block_groups[i].size
            ):
                new_free_block_groups = (
                    BlockGroup(None, block_groups[i].size, True),
                    BlockGroup(None, block_groups[j].size - block_groups[i].size, True),
                )
                if new_free_block_groups[1].size > 0:
                    block_groups[j] = new_free_block_groups[0]
                    block_groups.insert(j + 1, new_free_block_groups[1])
                    i += 1
                block_groups[j], block_groups[i] = block_groups[i], block_groups[j]
                break
    i -= 1

blocks = get_blocks(block_groups)
# print(blocks)

checksum = get_checksum(blocks)

end_time = datetime.now()

print("Part 2 Checksum:", checksum)
print("Part 2 executed in:", (end_time - start_time).total_seconds(), "seconds.")
````
